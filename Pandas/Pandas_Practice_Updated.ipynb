# Pandas Practice Demo

Basic exploratory data analysis using pandas.

```python
import pandas as pd

df = pd.read_csv('data/survey_results_public.csv')
df.head()
```

## Dataset Overview

```python
df.shape

df.info()
```

## Missing Values

```python
df.isna().sum().sort_values(ascending=False).head(10)
```

## GroupBy Example

```python
df.groupby('Country')['ConvertedComp'].mean().sort_values(ascending=False).head(10)
```

## Filtering Example

```python
df[df['ConvertedComp'] > 100000][['Country', 'ConvertedComp']].head()
```

## Average Missing Values Per Column (Sorted by Highest)

```python
(df.isna().mean() * 100).sort_values(ascending=False).head(15)
```

## Filtering

```python
high_earners = df[df['ConvertedComp'] > 100000]
high_earners[['Country', 'ConvertedComp']].head()
```
