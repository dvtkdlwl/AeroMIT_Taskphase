```python
#Completing imports:
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def rescaleFrame(frame, scale=0.4):
    height = int(frame.shape[0] * scale)
    width = int(frame.shape[1] * scale)

    dimensions = (width, height)

    return cv2.resize(frame, dimensions, interpolation = cv2.INTER_AREA)
```

```python
def getcontoursblack(img):
    # Finding all contours to store them in the contours variable:
    contours, hierarchy= cv2.findContours(img,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
    rectangles = []
    # Area Threshold:
    for contour in contours:
        if 200< cv2.contourArea(contour)<500:
            x, y, w, h = cv2.boundingRect(contour)
            rectangles.append((x, y, w, h))

    if len(rectangles)==0:
            cv2.putText(imgcontour, "0", (1000,100), cv2.FONT_HERSHEY_SIMPLEX, 1, (255,0,0), 2, cv2.LINE_AA, False)
    for rect in rectangles:
        x, y, w, h = rect
        cv2.rectangle(imgcontour, (x, y), (x + w, y + h), (255, 0, 0), 2)
        if len(rectangles)<20:
            cv2.putText(imgcontour, str(len(rectangles)), (1000,100), cv2.FONT_HERSHEY_SIMPLEX, 1, (255,0,0), 2, cv2.LINE_AA, False)
        else:
            cv2.putText(imgcontour, "NA", (1000,100), cv2.FONT_HERSHEY_SIMPLEX, 1, (255,0,0), 2, cv2.LINE_AA, False)
```

```python
def getcontoursred(img):
    # Finding all contours to store them in the contours variable:
    contours, hierarchy= cv2.findContours(img,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
    rectangles = []
    # Area Threshold:
    for contour in contours:
        if 400< cv2.contourArea(contour)<1000:
            x, y, w, h = cv2.boundingRect(contour)
            rectangles.append((x, y, w, h))
            
    if len(rectangles)==0:
            cv2.putText(imgcontour, "0", (900,100), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2, cv2.LINE_AA, False)
    for rect in rectangles:
        x, y, w, h = rect
        cv2.rectangle(imgcontour, (x, y), (x + w, y + h), (0, 0, 255), 2)
        if len(rectangles)<20:
            cv2.putText(imgcontour, str(len(rectangles)), (900,100), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2, cv2.LINE_AA, False)
        else:
            cv2.putText(imgcontour, "NA", (900,100), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2, cv2.LINE_AA, False)
```

```python
cap = cv2.VideoCapture('football_converted.mp4') 

while(True):
    # Capture frame-by-frame
    ret, frame = cap.read()
    resized = rescaleFrame(frame)
    imgcontour = resized.copy()
    #Drawing rectangle over the red subscribe button, the scoreboard and the Arsenal Logo.
    #This is helpful in avoiding false positives of the red team:
    cv2.rectangle(resized, (55,65), (380,100), (255,255,255), -1)
    cv2.rectangle(resized, (0,0), (1500,200), (255,255,255), -1)
    cv2.rectangle(resized, (0,600), (1500,1000), (255,255,255), -1)
    cv2.imshow('resizedframe', resized)
    hsv_image = cv2.cvtColor(resized, cv2.COLOR_BGR2HSV)

    blurred = cv2.blur(resized, (7,7))
    cv2.imshow('blurred', blurred)
    lower_black=np.array([48,0,0])
    upper_black=np.array([179,60,57])
    maskblack=cv2.inRange(blurred,lower_black,upper_black)
    kernel=np.ones((3,3),np.uint8)
    kernel1=np.ones((5,5),np.uint8)

    
    erodedblack = cv2.erode(maskblack, kernel, iterations =0)
    dilatedblack=cv2.dilate(erodedblack, kernel1 , iterations=2)
    cv2.imshow('maskblack',maskblack)
    cv2.imshow('erodedblack',erodedblack)
    cv2.imshow('dilatedblack',dilatedblack)
    #resblack=cv2.bitwise_and(hsv_image, hsv_image,mask=maskblack)
    #cv.imshow('res',resblack)

    lower_red=np.array([175,172,150])
    upper_red=np.array([179,255,210])
    maskred=cv2.inRange(hsv_image,lower_red,upper_red)

    kernel2=np.ones((3,3),np.uint8)
    erodedred = cv2.erode(maskred, kernel2, iterations =1)
    cv2.imshow('erodedred',erodedred)
    kernel2=np.ones((5,5),np.uint8)
    dilatedred=cv2.dilate(erodedred, kernel1 , iterations=5)
    cv2.imshow('maskred',maskred)
    cv2.imshow('dilatedred',dilatedred)
    
    getcontoursblack(dilatedblack)
    getcontoursred(dilatedred)

    # Display the final frame:
    cv2.imshow("ImgContour", imgcontour)
                   
    if cv2.waitKey(0) & 0xFF == ord('q'):
        break

# When everything done, release the capture
cap.release()
cv2.destroyAllWindows()
```
