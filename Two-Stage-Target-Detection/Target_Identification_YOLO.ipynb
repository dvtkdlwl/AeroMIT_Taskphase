```python
import cv2
from ultralytics import YOLO

model = YOLO("weights/best.pt")
conf_threshold = 0.5

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame, stream=False)

    annotated_frame = results[0].plot().copy()

    boxes = results[0].boxes
    high_conf = False
    if boxes is not None and len(boxes) > 0:
        confs = boxes.conf.cpu().numpy()
        if any(confs >= conf_threshold):
            high_conf = True

    if not high_conf:
        text = "Not a Target"
        font = cv2.FONT_HERSHEY_SIMPLEX  # OpenCV font
        font_scale = 1.2                 # slightly larger
        thickness = 2                     # bolder
        color = (0, 0, 255)              # red in BGR
        cv2.putText(annotated_frame, text, (10, 40), font,
                    font_scale, color, thickness, cv2.LINE_AA)

    cv2.imshow("YOLOv8 Live Camera", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
