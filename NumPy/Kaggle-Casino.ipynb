### pivot and pivot_table are used to reshape/reconstruct tables.
### They can convert columns to rows and vice versa. 
### The difference is that pivot cannot handle duplicates, while pivot_table can. The aggregate function in pivot_table can be of multiple types like mean/sum etc to handle the duplicate values mentioned previously.

```python
df2=df.iloc[0:5,0:5]
df2
```

```python
df2.pivot(columns= 'Fiscal Year')
```

### In the following example, pivot_table is used to handle duplicate value. I have used the sum function here, but other aggregate functions like mean can also be used.

## Pre-Processing Techniques:
Now for pre-processing techniques. I came across feature scaling, which is important primarily because a lot of algorithms like 'K-Means Clustering' rely on the Euclidean distance between two points. This Euclidean distance can become very large due to different units used to express magntitudes. For example: 
If height is expressed in cm, let's say 170 and 180cm respectively for two samples.
And if weight is expressed in kg, let's say 60 and 65kg respectively for two samples.
Then the corresponding data points become (170,60) and (180,65).

This means the Euclidean distance is pretty large. Why is this important? 
Well, because a lot of calculations like linear regressions become faster if Euclidean distances are smaller. In the case of linear regression, the convergence point is achieved much quicker if the values are scaled down because calculations become less intense.  
Also, CNN's take input as images whose pixel values can be between 0 to 255. To convert these into normalised ranges, techniques like normalisation can be used.

The most common library I come across is scikitlearn.preprocessing.

## Normalisation:

Normalisation is basically done to ensure all values are between 0 to 1. The maximum value becomes 1, while the minimum becomes 0.

How is this done?

The formula is: *x (normalised) = [x - x(min)]/ x(max) - x(min)*

## Standardisation:

It converts the entire distribution into a distribution with *Mean = 0* and *Standard Deviation = 1* . Note that it does NOT alter the table into a normalised bell curve, that is just a misconception- albeit a popular one.

## Imputation:

Imputation is used to fill out missing values from a dataset. It can be achieved using a variety of methods. We can calculate the mean of two points equidistant from it (before and after). Or we can also use ML models that predict the missing value based on other features for that record.

## Encoding Variables as Numbers:

A lot of times, values like 'high', 'medium' and 'low' can be encoded to 0,1,2 respectively to make operations easier. 
Further, NaN (Not a Number) values can be encountered in datasets to represent missing data. This data type cannot be operated upon, and so it makes sense to drop rows or columns if the number of NaN values in them are too high and/or these rows or columns are very less in comparison to the total dataset. We can also fill them with values like 0 or -1 to make calculations easier.  
Such operations are easily achievable by in-built PANDAS features like pd.dropna() or pd.fillna().
